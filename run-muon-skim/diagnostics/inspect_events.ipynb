# FSD Cube skim truth event display
Rotate/zoom the 3D plot; click legend entries to hide species. All coordinates are **EDepSim x,y,z in cm**, before larnd-sim swaps x and z; y is vertical. The box comes from the skim provenance. Segments are energy deposits, not a reconstructed trajectory fit.

The endpoint-active option is a geometry-only diagnostic. It does not apply larnd-sim neutral-particle, timing, or pixel-batch selection. A segment crossing the box with both endpoints outside is excluded by this mask. Detector view limits the viewport; it does not clip or modify the input segments.

```python
from pathlib import Path
import json
from event_display import load_event, load_bundle, save_bundle, summary, figure, TARGETS

CAMPAIGN = Path('/global/cfs/cdirs/dune/users/cborden/muon-skim-jobs/FSDCubeMuons_4500_v2')
FILE_INDEX, EVENT_ID = 62, 17
BUNDLE = None  # Or Path('/path/to/exported-event.hdf5') for an offline display.
ACTIVE_ONLY = False
DETECTOR_VIEW = True  # False shows the entire translated event.
print('Suggested (file, event) targets:', TARGETS)
```

```python
event = load_bundle(BUNDLE) if BUNDLE else load_event(CAMPAIGN, FILE_INDEX, EVENT_ID)
print(json.dumps(summary(event), indent=2))
print('Original input:', event['metadata']['source']['path'])
print('Translation [cm]:', event['metadata']['record']['shift_cm'])
```

```python
fig = figure(event, active_only=ACTIVE_ONLY, detector_view=DETECTOR_VIEW)
fig.show()
```
